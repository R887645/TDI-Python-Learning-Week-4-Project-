```python
#Task 1 : Assess Data Quality
import pandas as pd

# Step 1: Load your dataset
# Replace "supply_chain.csv" with the actual file name or path
df = pd.read_csv("supply_chain.csv")

# Step 2: Check for Missing Values
# This shows how many empty (NaN) values are in each column
print("Missing values by column:")
print(df.isnull().sum())
```

```python
#Task2 : Clean String Data
import pandas as pd 
# Step 1: Load your dataset 
df = pd.read_csv("supply_chain.csv") 
# Step 2: Identify text (string) columns # These are usually 'object' type in pandas 
text_columns = df.select_dtypes(include='object').columns 
print("Text columns found:", list(text_columns))
```

```python
# Step 3: Clean each text column 
for col in text_columns: 
# Remove leading/trailing spaces 
    df[col] = df[col].str.strip() 
    
    # Standardize capitalization (choose one style) 
    # Here we use title case (first letter capitalized) 
    df[col] = df[col].str.title() 
    
    # Fix common misspellings or variations 
    # Example: replace "Moter" with "Motor" 
    df[col] = df[col].replace("Moter", "Motor", regex=True)
```

```python
# Step 4: Document changes 
# Print a simple report of what we did print("\n=== String Data Cleaning Report ===") 
print("1. Removed leading and trailing spaces") 
print("2. Standardized capitalization to Title Case") 

print("3. Fixed common misspellings (e.g., 'Moter' → 'Motor')")

# Step 5: Save cleaned data to a new file 
df.to_csv("supply_chain_cleaned.csv", index=False) 
print("\nCleaned data saved to 'supply_chain_cleaned.csv'")
```

```python
#Task 3 : Handling Missing Values
import pandas as pd

# Step 1: Load your dataset
df = pd.read_csv("supply_chain.csv")

# Step 2: Find columns with missing values
print("Missing values by column:")
print(df.isnull().sum())
```

```python
# Step 3: Decide what to do for each column
# Example decisions (you can adjust based on your dataset):
# - Drop rows if the missing values are very few and not critical
# - Fill with a default value if it makes sense (e.g., 0 for Quantity, "Unknown" for Supplier)
# - Leave as is if the missing values are acceptable

# Example handling:
# If 'Quantity' has missing values, fill them with 0
if 'Quantity' in df.columns:
    df['Quantity'] = df['Quantity'].fillna(0)
    print("Filled missing Quantity values with 0")

# If 'SupplierName' has missing values, fill them with "Unknown"
if 'SupplierName' in df.columns:
    df['SupplierName'] = df['SupplierName'].fillna("Unknown")
    print("Filled missing SupplierName values with 'Unknown'")

# If 'DeliveryDate' has missing values, drop those rows (dates are critical)
if 'DeliveryDate' in df.columns:
    df = df.dropna(subset=['DeliveryDate'])
    print("Dropped rows with missing DeliveryDate")

# Step 4: Document decisions
print("\n=== Missing Values Handling Report ===")
print("1. Quantity: filled missing values with 0 (assume no items)")
print("2. SupplierName: filled missing values with 'Unknown'")
print("3. DeliveryDate: dropped rows because delivery date is essential")

# Step 5: Save cleaned dataset
df.to_csv("supply_chain_missing_cleaned.csv", index=False)
print("\nCleaned dataset saved to 'supply_chain_missing_cleaned.csv'")
```

```python
#Task 4: Remove Duplicates
import pandas as pd

# Step 1: Load your dataset
df = pd.read_csv("supply_chain.csv")

# Step 2: Find duplicate order_ids
duplicate_order_ids = df[df.duplicated(subset=['order_id'], keep=False)]

print("Duplicate order_ids found:")
print(duplicate_order_ids['order_id'].value_counts())


```

```python
# Step 3: Check if duplicates are true duplicates or legitimate repeated orders
# True duplicates = rows that are completely identical
true_duplicates = df[df.duplicated(keep=False)]

# Legitimate repeats = same order_id but different details (e.g., different product or quantity)
legit_repeats = duplicate_order_ids[~duplicate_order_ids.duplicated(keep=False)]

print("\nNumber of true duplicate rows:", true_duplicates.shape[0])
print("Number of legitimate repeated orders:", legit_repeats.shape[0])

# Step 4: Handle duplicates
# Remove true duplicates but keep legitimate repeats
df_cleaned = df.drop_duplicates()

# Step 5: Document what was found
print("\n=== Duplicate Records Report ===")
print("1. Found duplicate order_ids:", duplicate_order_ids['order_id'].nunique())
print("2. True duplicates removed:", true_duplicates.shape[0])
print("3. Legitimate repeats kept:", legit_repeats.shape[0])

# Step 6: Save cleaned dataset
df_cleaned.to_csv("supply_chain_no_duplicates.csv", index=False)
print("\nCleaned dataset saved to 'supply_chain_no_duplicates.csv'")
```

```python
#Task 5: Detect and document errors
import pandas as pd

# Step 1: Load your dataset
df = pd.read_csv("supply_chain.csv")

# Step 2: Detect logical errors
# Negative quantities (should always be positive)
negative_quantities = df[df['Quantity'] < 0] if 'Quantity' in df.columns else pd.DataFrame()

# Delivery dates before order dates (should not happen)
delivery_before_order = (
    df[df['DeliveryDate'] < df['OrderDate']]
    if 'DeliveryDate' in df.columns and 'OrderDate' in df.columns
    else pd.DataFrame()
)

# Lead times that don't match actual delivery times
# Example: LeadTime column should equal (DeliveryDate - OrderDate).days
lead_time_errors = pd.DataFrame()
if 'LeadTime' in df.columns and 'OrderDate' in df.columns and 'DeliveryDate' in df.columns:
    # Calculate actual lead time
    df['ActualLeadTime'] = (pd.to_datetime(df['DeliveryDate']) - pd.to_datetime(df['OrderDate'])).dt.days
    lead_time_errors = df[df['LeadTime'] != df['ActualLeadTime']]

# Other logical impossibilities (example: Price <= 0)
impossible_prices = df[df['Price'] <= 0] if 'Price' in df.columns else pd.DataFrame()

# Step 3: Create error report
print("\n=== Logical Errors Report ===")

print("1. Negative quantities:", negative_quantities.shape[0])
print("2. Delivery dates before order dates:", delivery_before_order.shape[0])
print("3. Lead time mismatches:", lead_time_errors.shape[0])
print("4. Impossible prices (<= 0):", impossible_prices.shape[0])

# Step 4: Document handling decisions
print("\n=== Handling Decisions ===")
print("1. Negative quantities: set to 0 or drop rows (depending on business rules).")
print("2. Delivery before order: drop rows (invalid).")
print("3. Lead time mismatches: recalculate lead time based on dates.")
print("4. Impossible prices: set to NaN or drop rows (depending on context).")

# Step 5: Apply fixes (example handling)
# Fix negative quantities by setting them to 0
if not negative_quantities.empty:
    df.loc[df['Quantity'] < 0, 'Quantity'] = 0

# Drop rows with delivery before order
if not delivery_before_order.empty:
    df = df.drop(delivery_before_order.index)

# Recalculate lead times
if not lead_time_errors.empty:
    df['LeadTime'] = df['ActualLeadTime']

# Drop rows with impossible prices
if not impossible_prices.empty:
    df = df.drop(impossible_prices.index)

# Step 6: Save cleaned dataset
df.to_csv("supply_chain_errors_cleaned.csv", index=False)
print("\nCleaned dataset saved to 'supply_chain_errors_cleaned.csv'")
```

```python
#Extra challenges
#Chalenge 1 : Write a function that validates a single order record. It should check all fields for type, 
#format, and logical validity. Return True if valid, False if invalid, with a message 
#explaining the problem.

import pandas as pd

def validate_order(order):
    """
    Validate a single order record.
    Input: order (a dictionary with keys like 'order_id', 'Quantity', 'Price', 'OrderDate', 'DeliveryDate', 'SupplierName')
    Output: (True/False, message)
    """

    # 1. Check order_id
    # It should be either a number (int) or text (str)
    if not isinstance(order.get('order_id'), (int, str)):
        return False, "order_id must be a number or text"

    # 2. Check Quantity
    # It should be a number and not negative
    if not isinstance(order.get('Quantity'), (int, float)):
        return False, "Quantity must be a number"
    if order['Quantity'] < 0:
        return False, "Quantity cannot be negative"

    # 3. Check Price
    # It should be a number greater than 0
    if not isinstance(order.get('Price'), (int, float)):
        return False, "Price must be a number"
    if order['Price'] <= 0:
        return False, "Price must be greater than 0"

    # 4. Check Dates
    # DeliveryDate should come after OrderDate
    try:
        order_date = pd.to_datetime(order.get('OrderDate'))
        delivery_date = pd.to_datetime(order.get('DeliveryDate'))
        if delivery_date < order_date:
            return False, "DeliveryDate cannot be before OrderDate"
    except Exception:
        return False, "OrderDate or DeliveryDate is not in a valid format"

    # 5. Check SupplierName
    # It should be text and not empty
    if not isinstance(order.get('SupplierName'), str) or order['SupplierName'].strip() == "":
        return False, "SupplierName must be a non-empty string"

    # If everything is fine
    return True, "Order record is valid"


# -------------------------------
# Example usage
# -------------------------------
order_record = {
    "order_id": 101,
    "Quantity": 5,
    "Price": 20.5,
    "OrderDate": "2024-01-10",
    "DeliveryDate": "2024-01-15",
    "SupplierName": "Motor Corp"
}

valid, message = validate_order(order_record)
print(valid, message)
```

```python
#Challenge 2: Create a before/after comparison showing data quality metrics (missing value count, 
#duplicate count, error count) from the original dataset and your cleaned dataset.
import pandas as pd

# Step 1: Load original dataset
df_original = pd.read_csv("supply_chain.csv")

# Step 2: Load cleaned dataset (from your previous cleaning steps)
df_cleaned = pd.read_csv("supply_chain_cleaned.csv")

# Step 3: Define a function to calculate metrics
def data_quality_metrics(df):
    metrics = {}
    # Count missing values
    metrics['Missing Values'] = df.isnull().sum().sum()
    # Count duplicate rows
    metrics['Duplicate Rows'] = df.duplicated().sum()
    # Count logical errors (examples)
    metrics['Negative Quantities'] = (df['Quantity'] < 0).sum() if 'Quantity' in df.columns else 0
    if 'OrderDate' in df.columns and 'DeliveryDate' in df.columns:
        metrics['Delivery Before Order'] = (pd.to_datetime(df['DeliveryDate']) < pd.to_datetime(df['OrderDate'])).sum()
    else:
        metrics['Delivery Before Order'] = 0
    metrics['Impossible Prices'] = (df['Price'] <= 0).sum() if 'Price' in df.columns else 0
    return metrics

# Step 4: Calculate metrics for both datasets
original_metrics = data_quality_metrics(df_original)
cleaned_metrics = data_quality_metrics(df_cleaned)

# Step 5: Create a comparison table

comparison = pd.DataFrame({
    "Original Dataset": original_metrics,
    "Cleaned Dataset": cleaned_metrics
})

print("\n=== Before/After Data Quality Comparison ===")
print(comparison)
```

```python
#Challenge 3: Identify patterns in the dirty data. Which columns have the most issues? Which 
#suppliers have the worst data quality? Which time periods had worse data entry? 
#Document your findings.
import pandas as pd

# Step 1: Load your original dataset
df = pd.read_csv("supply_chain.csv")

# Step 2: Check issues by column
issues_by_column = {}

# Missing values per column
issues_by_column['Missing Values'] = df.isnull().sum()

# Duplicate rows (count per column not possible, but overall count)
issues_by_column['Duplicate Rows'] = df.duplicated().sum()

# Negative quantities
if 'Quantity' in df.columns:
    issues_by_column['Negative Quantities'] = (df['Quantity'] < 0).sum()

# Impossible prices
if 'Price' in df.columns:
    issues_by_column['Impossible Prices'] = (df['Price'] <= 0).sum()

# Delivery before order
if 'OrderDate' in df.columns and 'DeliveryDate' in df.columns:
    issues_by_column['Delivery Before Order'] = (
        pd.to_datetime(df['DeliveryDate']) < pd.to_datetime(df['OrderDate'])
    ).sum()

print("\n=== Issues by Column ===")
for issue, counts in issues_by_column.items():
    print(issue, ":", counts)


# Step 3: Check issues by supplier
if 'SupplierName' in df.columns:
    supplier_issues = df.groupby('SupplierName').apply(
        lambda x: (x['Quantity'] < 0).sum() +
                  (x['Price'] <= 0).sum() +
                  ((pd.to_datetime(x['DeliveryDate']) < pd.to_datetime(x['OrderDate'])).sum()
                   if 'OrderDate' in x.columns and 'DeliveryDate' in x.columns else 0)
    )
    print("\n=== Issues by Supplier ===")
    print(supplier_issues.sort_values(ascending=False))

# Step 4: Check issues by time period
if 'OrderDate' in df.columns:
    df['OrderDate'] = pd.to_datetime(df['OrderDate'])
    issues_by_month = df.groupby(df['OrderDate'].dt.to_period("M")).apply(
        lambda x: (x['Quantity'] < 0).sum() +
                  (x['Price'] <= 0).sum() +
                  ((pd.to_datetime(x['DeliveryDate']) < pd.to_datetime(x['OrderDate'])).sum()
                   if 'DeliveryDate' in x.columns else 0)
    )
    print("\n=== Issues by Time Period (Month) ===")
    print(issues_by_month.sort_values(ascending=False))
```
